## Eksperiment med forskellig antalsparameter og sandsynlighedsparameter

Vi vil undersøge hvordan fordelingerne ændrer sig når antalsparameteren og sandsynlighedsparameteren ændrer sig.

````python
X = []
Y = []

def bin(n, p):
    x = 0
    for i in range(0, n):
        r = random.random()
        if r <= p:
            x = x + 1
    return x / n  # returnerer frekvensen i stedet for antal

# her genereres 100 realiseringer af to binominalfordelte variable X og Y med
# forskellig antalsparameter og sandsynlighed bin(n,p)
for k in range(0, 100):
    X.append(bin(10, 0.5))
    Y.append(bin(100, 0.5))

# Beregninger
meanfX = sum(X) / len(X)
varfX = sum(((x - meanfX)**2 for x in X)) / len(X)
stdfX = varfX**0.5

meanfY = sum(Y) / len(Y)
varfY = sum(((y - meanfY)**2 for y in Y)) / len(Y)
stdfY = varfY**0.5

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(range(1, len(X) + 1), X, 'o', label='n=10')
plt.plot(range(1, len(Y) + 1), Y, 'ro', label='n=100')
plt.xlabel('Forsøg nr.')
plt.ylabel('Frekvens')
plt.legend()
plt.grid(True)
plt.show()

# Output
print(f"X (n=10):")
print(f"  Middelværdi: {meanfX:.4f}")
print(f"  Spredning: {stdfX:.4f}")

print(f"\nY (n=100):")
print(f"  Spredning: {stdfY:.4f}")
print(f"  Middelværdi: {meanfY:.4f}")
````

## Sammenligning med teoretiske værdier

For at finde middelværdien og spredningen bruger vi data og disse definitioner.

### Middelværdien
af en stokastisk variabel *X* betegnes $\mu$, og udregnes som det vægtede gennemsnit:
$$\mu = p_1 x_1 + p_2 x_2 + p_3 x_3 + \dots + p_n x_n$$

### Varians
Var(*X*), af en stokastisk variabel *X* med middelværdien $\mu$ repræsenterer det gennemsnitlige afstandskvadrat til middelværdien $\mu$. Det udregnes som:
$$\text{Var}(X) = p_1(x_1 - \mu)^2 + \dots + p_n(x_n - \mu)^2$$

### Spredningen
$\sigma$, udregnes som kvadratroden af variansen:
$$\sigma = \sqrt{\text{Var}(X)}$$

### Teoretiske værdier for binomialfordelingen
Hvis *X* er binomialfordelt kan vi også udregne middelværdi og spredning teoretisk. De er givet ved

**Middelværdi for binomialfordelingen:**
$$\mu = p \cdot n$$
hvor *n* er antalsparameteren, og *p* er sandsynlighedsparameteren.

**Varians og spredning for binomialfordelingen:**
Hvis *X* er binomialfordelt med antalsparameter *n* og sandsynlighedsparameter *p*, så er varians og spredning for *X* givet ved:
$$\text{Var}(X) = np(1 - p)$$
$$\sigma(X) = \sqrt{\text{Var}(X)} = \sqrt{n \cdot p \cdot (1-p)}$$

## Øvelse: Sammenligning af empiriske og teoretiske værdier

````python
X = []

def bin(n, p):
    x = 0
    for i in range(0, n):
        r = random.random()
        if r <= p:
            x = x + 1
    return x

# her genereres 100 realiseringer af en binominalfordelt variabel X
# antalsparameter n og sandsynlighed p
n = 100
p = 0.5

for k in range(0, 100):
    X.append(bin(n, p))

# Beregninger - empiriske værdier
meanX = sum(X) / len(X)
varX = sum(((x - meanX)**2 for x in X)) / len(X)
stdX = varX**0.5

# Beregninger - teoretiske værdier
mu_teoretisk = p * n
var_teoretisk = n * p * (1 - p)
std_teoretisk = np.sqrt(var_teoretisk)

# Output
print("Observerede værdier:")
print(X)

print(f"\n{'':30} | {'Empirisk':<12} | {'Teoretisk':<12}")
print(f"{'-'*30}-+-{'-'*12}-+-{'-'*12}")
print(f"{'Middelværdi':30} | {meanX:<12.4f} | {mu_teoretisk:<12.4f}")
print(f"{'Varians':30} | {varX:<12.4f} | {var_teoretisk:<12.4f}")
print(f"{'Spredning':30} | {stdX:<12.4f} | {std_teoretisk:<12.4f}")

# Plotting - første plot
plt.figure(figsize=(10, 5))
plt.plot(range(0, len(X)), X, 'o')
plt.xlabel('Forsøg nr.')
plt.ylabel('Antal succeser')
plt.title(f'Binomialfordeling (n={n}, p={p})')
plt.grid(True)
plt.show()

# Plotting - sammenligning med teoretiske værdier
plt.figure(figsize=(10, 5))
plt.plot(range(0, len(X)), X, 'o', label='Observerede værdier')
plt.axhline(meanX, color='blue', linestyle='-', label=f'Empirisk middelværdi = {meanX:.2f}')
plt.axhline(mu_teoretisk, color='green', linestyle='--', label=f'Teoretisk middelværdi = {mu_teoretisk:.2f}')
plt.axhline(meanX + stdX, color='red', linestyle=':', label=f'μ ± σ (empirisk)')
plt.axhline(meanX - stdX, color='red', linestyle=':')
plt.axhline(mu_teoretisk + std_teoretisk, color='orange', linestyle=':', label=f'μ ± σ (teoretisk)')
plt.axhline(mu_teoretisk - std_teoretisk, color='orange', linestyle=':')
plt.xlabel('Forsøg nr.')
plt.ylabel('Antal succeser')

plt.legend()plt.show()
plt.grid(True)
````
