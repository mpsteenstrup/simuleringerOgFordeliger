# Random walk

Ved en random walk bevæger man sig tilfældigt frem eller tilbage.

Brownian motion er den tilfældige bevægelse som partikler foretager. Det kan beskrives som en random walk.

[Brownian motion Wikipedia](https://en.wikipedia.org/wiki/Brownian_motion#/media/File:Brownian_motion_large.gif)

I de første to simuleringer er en random walk genereret i to dimensioner ved at gå den samme længde i hvert skridt og dreje sig tilfældigt [0, 90, 180, 270] grader.

## Visualisering af random walk (2D med turtle)

**Note:** Denne simulation bruger turtle graphics som fungerer bedst i trinket.io. I Jupyter kan du se resultatet af simuleringen nedenfor.

```python
import random
from math import *
import matplotlib.pyplot as plt
import numpy as np
```

## Random walk med afstandsberegning (2D)

Her beregnes afstanden fra start efter et antal skridt.

```python
l = [0, pi/2, pi, pi*3/2]
i = 0
j = 0
x = 0
y = 0
d = []

# Generering af random walks
steps = 100
walks = 100

while j < walks:
    x = 0
    y = 0
    i = 0
    
    while i < steps:
        angle = random.choice(l)
        x += cos(angle)
        y += sin(angle)
        i += 1
    
    # Beregn afstand fra start
    distance = (x**2 + y**2)**0.5
    d.append(distance)
    j += 1

# Plotting
plt.plot(range(1, len(d) + 1), d, 'o')
plt.show()

# Output
print(d)
```

## Random walk 1D (hurtig version)

Hvis det skal gå lidt hurtigere er der ikke nogen grund til at se den blive tegnet. Her er den 1D version.

```python
l = [-1, 1]
i = 0
j = 0
x = 0
d = []

# Parametrer
steps = 1000
walks = 300

# Generering
while j < walks:
    x = 0
    i = 0
    while i < steps:
        x += random.choice(l)
        i += 1
    d.append(x)
    j += 1

# Plotting
plt.plot(range(1, len(d) + 1), d, 'o')
plt.show()

# Output
print(d)
```

## Random walk 2D (hurtig version med mange walks)

```python
l = [0, pi/2, pi, pi*3/2]
i = 0
j = 0
x = 0
y = 0
d = []

# Parametrer
steps = 1000
walks = 100

# Generering
while j < walks:
    x = 0
    y = 0
    i = 0
    while i < steps:
        angle = random.choice(l)
        x += cos(angle)
        y += sin(angle)
        i += 1
    
    distance = (x**2 + y**2)**0.5
    d.append(distance)
    j += 1

# Plotting
plt.plot(range(1, len(d) + 1), d, 'o')
plt.show()

# Output
print(d)
```
