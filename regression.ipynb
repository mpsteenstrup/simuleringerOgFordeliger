# Normalfordelte residualer

En måde at tjekke om man har brugt den rigtige model når man laver regression er ved at tjekke om residualerne er normalfordelte. Hvis det er tilfældet, kan man ikke finde en bedre model og det som er tilbage er rent tilfældighed.

I simuleringen bruger vi at summen af en binomialfordeling er normalfordelt. x og y data bliver udskrevet nedenfor.

```python
import matplotlib.pyplot as plt
import random
import numpy as np
```

```python
a = 2  # Slope
b = 5  # Intercept
i = 0
x = []
y = []

def bin(n, p):
    x = 0
    for i in range(0, n):
        r = random.random()
        if r <= p:
            x = x + 1
    return x / n

# Generering af data
while i < 100:
    x.append(i)
    y.append(a * i**2 + b + 100 * bin(10, 0.5))
    i += 1

# Konverter til numpy arrays
# Plotting
plt.plot(x, y, 'o')

# Output
print(x)
print(y)
```

## Øvelser

* Foretag lineær regression og undersøg om residualerne er normalfordelte.
* Prøv at lave om i modellen (ændr formlen for y).
* Sammenlign residualerne fra den lineære og polynomielle model.
* Hvilken model giver de mest normalfordelte residualer?

Når residualerne er normalfordelte kan man også bestemme konfidensintervaller for konstanterne i regressionen.

* Undersøg hvad usikkerheden er på a og b.
* Hvad siger testet om a-værdien?
* Test hypotesen H₀: a = 0. p-værdien giver sandsynligheden for at H₀ hypotesen er rigtig.

## Yderligere analyse

Du kan eksperimentere med forskellige modeller ved at ændre funktionen:

```python
# Eksempel: Prøv en kubisk model
a_new = 0.01
b_new = 2
c_new = 5

x_new = []
y_new = []

for i in range(100):
    x_new.append(i)
    # Ny model: y = a*x^3 + b*x + c + noise
    y_new.append(a_new * i**3 + b_new * i + c_new + 100 * bin(10, 0.5))

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(x_new, y_new, 'o')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Ny model til analyse')
plt.grid(True)
plt.show()

print("x:", x_new)
print("y:", y_new)
```
