# Triangeltest

Triangeltesten bruges bl.a. til at teste smag. Vi kan ikke teste om der er forskel på forskellige cola mærker, men vi kan teste hvorvidt vi kan smage forskel.

Ved testen undersøges tre prøver hvoraf to er ens. Testen går ud på at finde den som er forskellig.

Resultatet kan selvfølgeligt simuleres.

```python
import matplotlib.pyplot as plt
import random
import numpy as np
```

```python
X = []

def bin(n, p):
    x = 0
    for i in range(0, n):
        r = random.random()
        if r <= p:
            x = x + 1
    return x

# her genereres 1000 realiseringer af en binominalfordelte variable X
# antalsparameter n=30 og sandsynlighed p=1/3
for k in range(0, 1000):
    X.append(bin(30, 1/3))

# Plotting
plt.plot(range(1, len(X)), X, 'o')
plt.show()

# Output
print(X)
```
