```python
t = make_turtle(width=400, height=400, delay=0)
#t = Turtle(delay=0)
t.pen_down()
i = 0
n=1000
s=1
while i<n:
  #t.left(360*random.random())
  t.left(random.uniform(0,360))
  t.forward(s)
  i+=1
d = (t.x**2+t.y**2)**0.5
print(d)
print(s*n**0.5)
```

```python
t = Turtle(delay=0)
t.hide()  # Gør det hurtigere
t.pen_up()     # Hvis du bare vil have slutdata uden streger

def sim(n):
    for _ in range(n):
        t.left(random.uniform(0, 360))
        t.forward(4)
    return t.x

x = 0
m=10
n=10
for i in range(m):
    x+=sim(n)
print(x/m)
```
